```python
# include necessary libraries
import numpy as np
import pandas as pd
import matplotlib as plt
import seaborn as sb
```

```python
# load dataset
data = pd.read_csv('housing.csv')
```

```python
# convert price to millions
data['price(millions)'] = data['price'] / 1000000 

# log transformation of price
data['log_price'] = np.log(data['price'])   

# Encoding columns 
cols = ["mainroad", "guestroom", "basement", 
        "hotwaterheating", "airconditioning" , "prefarea"]

data[cols] = data[cols].replace({"no": 0 , "yes": 1})

# for selecting numeric data
numeric_data = data.select_dtypes(include=['float64', 'int64'])
```

```python
# price distribution
sb.histplot( data['price(millions)'], kde=True)
# scatter plot between price and area
sb.scatterplot(x='area', y='price(millions)', data=data)
# correlation between bedrooms and price
sb.barplot(x='bedrooms', y='price(millions)', data=data)
# box plot between furnishing status and price
sb.barplot( x='furnishingstatus', y='price(millions)', data=data)
# heatmap for numeric data

corr = numeric_data.corr()

sb.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
```
